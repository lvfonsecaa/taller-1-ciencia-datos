# 02 · Estrategia de análisis

A partir del entendimiento inicial de los datos, se considera que los principales atributos con los cuales se desarrollará el análisis son el valor del contrato, el sector, la modalidad de contratación, los días adicionados y la liquidación. Se utilizarán el valor del contrato, el sector y la modalidad de contratación para identificar si determinados rangos de valor, sectores o modalidades están asociados con una mayor ocurrencia de desviaciones. Asimismo, se analizarán los días adicionados y la ausencia de liquidación cuando corresponda, ya que representan dos de las desviaciones respecto al plan original de los contratos planteadas en el problema. Aunque estos atributos serán priorizados, también podrán incorporarse otras variables del conjunto de datos que resulten relevantes durante el análisis.

Para el desarrollo del análisis, primero se calcularán estadísticos descriptivos que permitan comparar las desviaciones entre diferentes grupos de contratos. Para esto, los contratos se agruparán según las categorías o rangos de los atributos analizados. Posteriormente, se utilizarán visualizaciones adecuadas que permitan identificar posibles relaciones y diferencias entre el comportamiento de los grupos. A partir de los patrones encontrados, se formularán hipótesis y se aplicarán las pruebas correspondientes para determinar si las diferencias observadas son estadísticamente significativas. Por ejemplo, se podrán comparar las proporciones de contratos que presentan una desviación según su modalidad de contratación. Finalmente, se propondrá una supervisión más cercana para los contratos que presenten las características que, según el análisis, estén asociadas con una mayor ocurrencia de desviaciones.
